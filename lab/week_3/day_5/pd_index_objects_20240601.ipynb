## Bonus 2

Players from Argentina are older for all positions in the data set compared to players from Nigeria. You are now asked to find out for which position shows the biggest age differenceand for which  which position shows the smallest age difference between the two countries.

Solve this using group by and unstack!

```python
my_df = df[["age"]].loc[["Nigeria","Argentina"]].groupby(["position","team"]).mean().unstack()
```

```python
my_df["age", "Argentina"] - my_df["age", "Nigeria"]

```
