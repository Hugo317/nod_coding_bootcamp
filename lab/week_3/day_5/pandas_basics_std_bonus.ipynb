```python
new_york = df.loc[df["State"] == "New York"]

california = df.loc[df["State"] == "California"]

texas = df.loc[df["State"] == "Texas"]

illinois = df.loc[df["State"] == "Illinois"]

my_lst = [new_york,california,texas,illinois]
```

```python
illinois
```

```python

# What is the most common desease in each of the four states.
```

```python
for i in my_lst:
    #common_dis = i["Disease"].value_counts().sort_values(ascending=False).iloc[0]
    common_dis = i.groupby(["Disease", "State"]).size().reset_index(name="Disease Count")
    print(common_dis)
    #i[["Disease","State"]].value_counts("Disease")
```
