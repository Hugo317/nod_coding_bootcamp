## Q1. Bubble Sort

#### What it is

Bubble Sort is a simple comparison-based sorting algorithm.
It repeatedly steps through the list, compares adjacent elements, and swaps them if they are in the wrong order.
Larger elements “bubble up” to the end of the list with each pass.

* Time complexity: O(n²)
* Space complexity: O(1) (in-place)
* Stable: Yes
* Practical use: Mostly educational (rarely used in production)

#### Steps to implement

1. Loop over the list n times.

2. On each pass:

    * Compare element i with element i + 1

    * Swap them if arr[i] > arr[i + 1]

3. After each pass, the largest unsorted element is guaranteed to be at the end.

4. (Optimization) Stop early if no swaps happen during a pass.

## Q2. Merge Sort

#### What it is

Merge Sort is a divide-and-conquer sorting algorithm.
It works by repeatedly splitting a list into smaller halves until each sublist contains one element, then merging those sublists back together in sorted order.

Key properties:

* Time complexity: O(n log n) (guaranteed)

* Space complexity: O(n) (not in-place)

* Stable: Yes

* Widely used in production (e.g. external sorting, large datasets)

#### Steps to implement

1. **Base case**:
If the list has 0 or 1 elements, it is already sorted.

2. **Divide**:
Split the list into two halves.

3. **Conquer**:
Recursively apply merge sort to each half.

4. **Merge**:
Combine the two sorted halves into a single sorted list by repeatedly taking the smallest front element.

## Q3. Quick Sort

#### What it is

Quick Sort is a divide-and-conquer sorting algorithm that works by selecting a pivot element and partitioning the list so that:

* elements smaller than the pivot come before it

* elements larger than the pivot come after it

It then recursively applies the same logic to the sublists.

#### Key properties

* Average time complexity: O(n log n)

* Worst-case time complexity: O(n²) (bad pivot choices)

* Space complexity: O(log n) (in-place, recursive)

* Stable: No

* Widely used in production due to excellent average-case performance

#### Steps to implement (in-place version)

1. Choose a pivot (commonly the last element).

2. Partition the array so that:

    * elements ≤ pivot are on the left

    * elements > pivot are on the right

3. Place the pivot in its correct sorted position.

4. Recursively apply quick sort to the left and right partitions.

```python
# Implement Quick Sort algorithm here:
def quick_sort(my_list):
    left_list = []
    right_list = []

    if len(my_list) > 1:
        pivot = my_list[-1]
        for i in my_list[:-1]:
            if i <= pivot:
                left_list.append(i)
            else:
                right_list.append(i)

        my_list = quick_sort(left_list) + [pivot] + quick_sort(right_list)
    

    return my_list  

    


```

## Q4. Binary Search

#### What it is

Binary Search is an efficient search algorithm for sorted data.
It repeatedly halves the search space by comparing the target value to the middle element.

#### Key properties:

* Time complexity: O(log n)

* Space complexity: O(1) (iterative)

* Requirement: Input must be sorted

* Widely used in production (databases, config lookups, feature flags)

#### Steps to implement (iterative)
1. Set two pointers:

    * `left` at the start of the list

    * `right` at the end of the list

2. While left <= right:
    * Compute `mid = (left + right) // 2`

    * If `arr[mid] == target, return mid`   

    * If `target < arr[mid], move right left`

    * If `target > arr[mid], move left right`

    * If the loop ends, the target is not in the list.

## Q5. Two Pointers

#### What it is

Two pointers is a technique (not one single algorithm) where you use two indices to scan through a sequence to solve problems more efficiently than nested loops.

#### Typical production use cases:

* Deduplicating sorted data

* Merging / comparing two sorted lists

* Finding pairs that match a condition (e.g., sum, difference)

* Interval and time-window processing

#### Common patterns:

1. Opposite ends (left and right move inward)

2. Same direction (fast/slow pointers)

3. Two sequences (one pointer per list)

#### Steps to implement (general)

1. Decide what the pointers represent (left/right, slow/fast, i/j on two lists).
