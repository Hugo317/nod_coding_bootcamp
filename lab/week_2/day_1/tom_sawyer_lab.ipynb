#### The next step is to create a function that will remove underscores. 

The underscores are there to make the text italic. However, since we just want the plain text, we are going to replace the underscores with an empty string. Write your function below.

```python
def underscore(x):
    return x.replace("_","")
    

    '''
    Input: A string
    Output: The string with underscores removed
    
    Example:
    Input: '_under_'
    Output: 'under'
    '''
    pass
```

```python
test_word = "___hugo"
test_word = underscore(test_word)
test_word
```

Now that we have our function, use the `map()` function to apply this function to our book, The Prophet. Return the resulting list to a new list called `sawyer_underscore`.

Another thing you may have noticed is that some words contain a line break. Let's write a function to split those words. Our function will return the string split on the character `\n`. Write your function in the cell below.

Have you noticed that some words contain a line break? Yeah, we want to get rid of those as well. You can do this by writing a function that splits on line breaks

```python
def line_breaks(x):
    return x.replace("\n","")
    '''
    Input: A string
    Output: A list of strings split on the line break (\n) character
        
    Example:
    Input: 'answer.\n\nThe'
    Output: ['answer', 'beloved']
    '''
    pass
```

Apply the `line_breaks` function to the `sawyer_underscore` list. Name the new list `sawyer_line`.

```python
sawyer_line = list(map(lambda x: line_breaks(x) ,tom_sawyer))
sawyer_line
```

# Challenge 3 - Reducing

#### Now that we have significantly cleaned up our text corpus, let's use the `reduce()` function to put the words back together into one long string separated by spaces. 

We will start by writing a function that takes two strings and concatenates them together with a space between the two strings.

```python
def concat_space(a, b):
    return a + " " + b
    
```

Use the function above to reduce the text corpus in the list `sawyer_filter` into a single string. Assign this new string to the variable `sawyer_string`.

```python
sawyer_string = reduce(concat_space,sawyer_filter)
sawyer_string
```
