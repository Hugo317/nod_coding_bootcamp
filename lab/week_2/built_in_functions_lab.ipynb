## 2.3

In this question we are doing almost the same thing we did in the previous question but we want to store the functionality in a user defined function. The point of doing this is that when we have stored a functionality in a user defined function, it is very efficient to apply our function to any variable (a string variable in this case, since we are counting word length inside a string)

Below are 3 strings that you can use separately to call on your function to try it out. 

**Example**

Call the function using: *function_name*(phrase3)

output: [1,2,3]

```python
phrase3 = "I am god"
phrase4 = "I am not god"
phrase5 = "Dont worry be happy said the fish"

def count_words(*args):
    return_lst = []
    if len(args) == 1:
        for i in args[0].split():
            return_lst.append(len(i))
    else:


        for i in args:
            mini_list = []
            for j in i.split():
                mini_list.append(len(j))
            return_lst.append(mini_list)

    
    return return_lst

count_words(phrase3)
```

## 2.3.2

Now do the same thing as in the previous question but instead of a list the function should return a set

Compare results for `phrase3` and `phrase4` from the two different user defined functions

```python
def count_words_set(*args):
    return_set = set()
    if len(args) == 1:
        for i in args[0].split():
            return_set.add(len(i))
    else:


        for i in args:
            mini_set = set()
            for j in i.split():
                mini_set.add(len(j))
            return_set.add(mini_set)

    
    return return_set
count_words_set(phrase3)
```

# Bonus challenge 2
Create a function that accepts 2 lists of equal length and a separator. The function should return one list that contains the items from both lists which are separated by the separator

**Example** 

lst1 = [A, B]\
lst2 = [a, b]\
separator = "-"

output: [A-a, B-b]

```python
lst1 = ["A","B","C"]
lst2 = ["a","b","c"]
separator = "-"
test = "a "
test = test.join("a").join("b")
print(test)
```

```python
def joiner_with_separator(lst1,lst2,sep):
    return [i + sep + lst2[index] for index, i in enumerate(lst1)]

joiner_with_separator(lst1,lst2,separator)
```

# Bonus challenge 3

Take a list of digits and return the number that they
correspond to. For example, \[1,2,3] corresponds to one-hundred-twenty-three (123)

*Do not convert the integers to strings!* 

To achieve this without converting the integers to strings we can make use of the `reduce` functionality

```python
from functools import reduce
digits = [3,4,3,2,1]
```

```python
#3      4       3       2       1
#x      x       x       x       x
#10k.   1k.     100.    10.     1. 
#10^4.  10^3.   10^2.   10^1.   10^0
```

```python
reduce(lambda x, y: x * 10 + y, digits)
```
