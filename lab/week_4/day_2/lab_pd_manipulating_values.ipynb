# Manipulating values the right way

Import pandas and read the `cars.csv` file

1. Create a subset of the DataFrame cars with all cars having an mpg higher than 40

    - Save the subset in a variable named `high_eff`
    - `high_eff` should be a new and independent DataFrame. Any changes in `high_eff` should not influence the original DataFrame

2. Cap the mpg for all cars in high_eff to 40 (Overwrite the mpg values).

    - Your code should not cause any warning message

3. Check in the original Dataframe whether there are still values greater than 40 in the "mpg" column

    - If so, your code is correct!
    - If not, find out what you did wrong and fix it!
