## Cleaning Area

```python
def matching_adresses(df):

    ### string cleaner

    def clean_string(to_clean):
        for index,i in enumerate(list(to_clean)):
            try:
                int(i)
            except:
                continue
            return to_clean[:index-1]

    ### getting no area df 

    no_area_df = df.loc[df["area"].isna()][["area","address"]]
    no_area_df["address"] = no_area_df["address"].apply(clean_string)

    ### creating with area df

    with_area = df[["address","area"]].drop(no_area_df.index)
    with_area["address"] = with_area["address"].apply(clean_string)




    index_to_drop = []
    for index, _ in enumerate(range(0, (len(no_area_df)))):
        if no_area_df.iloc[index]["address"] in with_area["address"].values:
            df.loc[no_area_df.index[index], "area"] = with_area.loc[
                with_area["address"] == no_area_df.iloc[index]["address"]
                ].iloc[0]["area"]
        else:
            index_to_drop.append(no_area_df.index[index])

    df.drop(index_to_drop,inplace=True)

    return df

df = matching_adresses(df)

       # clean_string(no_area_lst.iloc[0]["address"]).isin(with_area.)


```

## FIXING DATE SOLD 

```python
df["date_sold"] = df["date_sold"].str.replace("Såld ", "")
df.date_sold
```

```python
month_map = {
    " januari ": "/1/", " februari ": "/2/", " mars ": "/3/", " april ": "/4/",
    " maj ": "/5/", " juni ": "/6/", " juli ": "/7/", " augusti ": "/8/",
    " september ": "/9/", " oktober ": "/10/", " november ": "/11/", " december ": "/12/"
}

df["date_sold"] = df["date_sold"].replace(month_map, regex=True)
```
