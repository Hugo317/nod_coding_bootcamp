#### 1.1. What genre of games have been made the most?

Visualize it ranging from most to least made.

````python
sns.countplot(data=df,x="Genre",order=df["Genre"].value_counts().index)
plt.xticks(rotation=45)
````

#### 1.2. Which year had the most game release?
Sort the result descendingly and make sure to rotate the ticks on the x-axis so that they don't overlap.

````python
sns.countplot(data=df,x="Year",order=df["Year"].value_counts(ascending=False).index)
plt.xticks(rotation=90)
plt.show()
````

#### 1.3. Plot the top 5 years in terms of releases, by genre.

Make sure you put the legend outside of the figure so that it doesn't cover the bars. You can use the `bbox_to_anchor` parameter to adjust. More info on different variation are found on this [link](https://pythonguides.com/put-legend-outside-plot-matplotlib/).

````python
df.loc[df["Genre"] == "Action"].groupby("Year").size().rename("games_count").sort_values(ascending=False).reset_index().head(5)
````

````python
for genre in df.Genre.unique():
    new_data = df.loc[df["Genre"] == genre].groupby("Year").size().rename("games_count").sort_values(ascending=False).head(5)
    print("head" , new_data.head(10))
    sns.barplot( x=new_data.index, y=new_data.values, order=new_data.index)
    plt.show()
````
