Import seaborn, matplotlib and pandas

Read the `countries_un.csv` file

```python
import pandas as pd
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
df = pd.read_csv("countries_un.csv")
```

# q1
Visualize the number of countries in each region

You should customize the figure so it looks reasonably appealing

```python
region_counts = df["Region"].value_counts()

plt.figure(figsize=(10, 6))
region_counts.plot(kind="bar")

plt.title("Number of Countries in Each Region")
plt.xlabel("Region")
plt.ylabel("Number of Countries")
plt.xticks(rotation=45)
plt.show()
```

# q2

Visualize the distributions of:
- literacy
- infant mortality rate

```python
sns.histplot(data=df, x="Literacy (%)")
plt.show()
```

```python
sns.histplot(data=df, x="Infant mortality (per 1000 births)")
plt.show()
```

```python
sns.barplot(data=df, x="Region" ,y="Literacy (%)")
plt.xticks(rotation=45)
plt.show()
```

```python
sns.barplot(data=df, x="Region" ,y="Infant mortality (per 1000 births)")
plt.xticks(rotation=45)
plt.show()
```

# q3

Visualize the average value for the different regions for the following columns: 
- Population	
- Pop. Density (per sq. mi.)	
- Literacy (%)

```python
df["Region"].values
df.columns
```

```python
for data in ["Population", "Pop. Density (per sq. mi.)", "Literacy (%)"]:
    sns.barplot(data=df, x="Region", y=data, estimator="mean",errorbar=None)
    plt.xticks(rotation=45, ha="right")
    plt.title(f"Average {data} by Region")
    plt.show()
```

# q4 

Visualize the relationship between Population & Area (sq. mi.) in a scatterplot. Specify the hue parameter with Region

```python
sns.scatterplot(data=df,x="Population",y="Area (sq. mi.)",hue="Region")
plt.legend(bbox_to_anchor=(0.9, -0.1), ncol=2)
```
