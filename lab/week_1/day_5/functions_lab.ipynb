## delete duplicates
Create a function `delete_duplicates` that takes a list as argument and removes all duplicate values.



````python
test_list = [2,11,2,3,4,5,6,2,3,4]
def list_delete_dups(input_list,reversed=False):
    if isinstance(input_list, list):
        
        return sorted(list(set(input_list)), reverse = reversed)
    else:
        print("Object not a list")

list_delete_dups(test_list,reversed=True)
````

If the second is greater, it should `"Second is greater"`. If they are equal to each other, it should return "`Nums are equal`".

````python
def compare_numbers(num_a,num_b):
    try:
        num_a = int(num_a)
        num_b = int(num_b)
        if num_a > num_b:
            return "First is greater!"
        elif num_a < num_b:
            return "Second is greater!"
        else:
            return "They are equal!"
    except:
        return "Both values should be numbers"

print(compare_numbers(20,10))
print(compare_numbers(2,10))
print(compare_numbers(10,10))
print(compare_numbers([],"LOL"))
````

## count letter

Write a function `count_letter`. The function takes two strings as parameters, the first one being a word and the second one being a letter. The function should return an integer that is equal to the number of times that the letter appears in the word. If the letter does not exist in the word the function should return 0. Make sure that you lowercase the word so that it is case insensitive.  


````python
def count_letter(input_str,input_letter,case_sensitive = True):
    if not isinstance(input_str,str):
        print("raise Invalid sintax only trings here")
    elif not isinstance(input_letter , str) or len(input_letter) > 1 or len(input_letter) < 1:
        print("raise Invalid sintax should be a single character")
    else:
        if case_sensitive:
            return input_str.count(input_letter)
        else:
            return input_str.lower().count(input_letter.lower())

print(count_letter("AHAHAHAHAH","a",case_sensitive=False))
````
