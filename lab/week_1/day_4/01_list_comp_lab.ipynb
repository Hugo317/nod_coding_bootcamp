# 4 - Nested loops to access each value in a 3D-matrix

Now we want to access each item in the 3D-matrix `b` (which is a 5 x 2 x 3 Numpy array)

```python
b = np.array([[[0.55867166, 0.06210792, 0.08147297],
        [0.82579068, 0.91512478, 0.06833034]],

       [[0.05440634, 0.65857693, 0.30296619],
        [0.06769833, 0.96031863, 0.51293743]],

       [[0.09143215, 0.71893382, 0.45850679],
        [0.58256464, 0.59005654, 0.56266457]],

       [[0.71600294, 0.87392666, 0.11434044],
        [0.8694668 , 0.65669313, 0.10708681]],

       [[0.07529684, 0.46470767, 0.47984544],
        [0.65368638, 0.14901286, 0.23760688]]])
```

## 4.1

Create a flat list containing each item in `b`

First do this without using list comprehension

If done correctly you should have a 1D-list containing 30 items

```python
list_ex4 = []
for i in b:
    for j in i:
        for k in j:
            list_ex4.append(k)
len(list_ex4)
```

## 4.2

Same objective as the previous question but now using list comprehension


```python
comp_list_ex_4 = [k for i in b for j in i for k in j]
len(comp_list_ex_4)
```

## 4.3
Now add a condition to the list comprehension you created in Q4.2 so that the new list only contains values **smaller than 0.5**


```python
comp_list_ex_4_1con = [k for i in b for j in i for k in j if float(k) > 0.5 ]
comp_list_ex_4_1con
```

## 4.4

Now add another condition to the list comprehension so that the new list only contains values **smaller than 0.5 and bigger than 0.1**

```python
comp_list_ex_4_2con = [k for i in b for j in i for k in j if float(k) < 0.5 and float(k) > 0.1 ]
```
