# Dictionary Comprehensions

A dictionary comprehension is a concise way to create a new dictionary in Python using a single line of code. It is similar to a list comprehension, but instead of creating a new list, it creates a new dictionary by iterating over an iterable and transforming the values using a key-value pair.

The syntax of a dictionary comprehension is as follows:
```python
new_dict = {key_expression: value_expression for item in iterable if condition}
```
Here, `key_expression` and `value_expression` are expressions that determine the key and value of each item in the new dictionary, `iterable` is an iterable object to loop over, and condition is an optional condition to filter items.

A dictionary comprehension is useful when you need to create a new dictionary from an existing iterable object, and you want to do it in a concise and readable way. It can also be used to transform an existing dictionary into a new dictionary by applying a function or expression to the keys or values.

For example, suppose you have a list of numbers, and you want to create a new dictionary where each key is a number in the list, and the value is the square of that number. You could use a dictionary comprehension to accomplish this task in a single line of code:

```python
numbers = [1, 2, 3, 4, 5]
squares = {num: num**2 for num in numbers}
print(squares)  # Output: {1: 1, 2: 4, 3: 9, 4: 16, 5: 25}
```
Here, the dictionary comprehension `{num: num**2 for num in numbers}` creates a new dictionary where each key is a number from the numbers list, and the value is the `square` of that `number`. The resulting dictionary is assigned to the squares variable and printed to the console.

Dictionary comprehensions are a nice feature of Python because they allow you to write concise and readable code that is easy to understand and maintain. They can also be used in combination with other Python features, such as functional programming and object-oriented programming, to create powerful and flexible solutions to complex problems.

Your task is to use a dictionary comprehension to create a new dictionary that only includes the names and grades of students who passed the class.


````python
grades = {'Alice': 80, 'Bob': 50, 'Charlie': 70, 'Dave': 90, 'Eve': 55, 'Frank': 65}
````
