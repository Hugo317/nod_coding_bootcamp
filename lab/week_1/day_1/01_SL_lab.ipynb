# SL: Passenger Control System

You travel from Ropsten to T-centralen to attend your new job as a Data Analyst on Tegelbacken. The subway that you are travelling with has a new entry and exit control system that monitors the passengers. The system has been implemented to detect if passenger capacity is to high.

Every stop from Ropsten to TC is represented by a tuple that consists of two integers. It looks like this:
```
ropsten = (in, out)
```
The succession of stops is represented by a list of these tuples.
```
stops = [(ropsten_in, ropsten_out), (gärdet_in, gärdet_out), (karlaplan_in, karlaplan_out), 
         (öt_in, öt_out), (tc_in, tc_out)]
```

````python
# Run this cell

stops = [(42, 0), (6, 4), (21, 14), (31, 34), (40, 56)]
stops_name = ["Ropsten","Gärdet","Karlaplan","öt","tc"] 
print(stops[1][1])
````

#### 1. Calculate the number of stops

````python
number_of_stops = len(stops)
print(number_of_stops)
````


#### 2. Calculate the occupation for every stop. Remember that each item is dependant on the preceeding item. 


````python
occupation = 0
stop_number = 0
for i in stops:
    occupation += i[0] - i[1]
    print(f"occupation at stop {stops_name[stop_number]} is {occupation}")
    stop_number += 1
       
````

#### 3. Find the maximum occupation of the bus.

````python
occupation = 0
max_occupancy = 0
for i in stops:
    occupation += i[0] - i[1]
    if occupation > max_occupancy: max_occupancy = occupation

print(max_occupancy)
````

#### 4. Calculate the average occupation and the standard deviation.

````python
#average
occupation = 0
occupation_adition = 0
number_of_stops = 0
for i in stops:
    occupation += i[0] - i[1]
    occupation_adition += occupation
    number_of_stops += 1 

print(occupation_adition / number_of_stops)
````

````python
import statistics
occupation = 0
occupation_list = []
for i in stops:
    occupation += i[0] - i[1]
    occupation_list.append(occupation)
print(occupation_list)
print(statistics.stdev(occupation_list))
````
