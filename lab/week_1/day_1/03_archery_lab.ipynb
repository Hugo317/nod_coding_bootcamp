# Nod Archery Championship! 

Today, Nod arranges their yearly competion - Nod Archery Championship! 

You are one of the favourite rookies this year and are aiming for a win. With your bow and arrow you will try to hit the bullseye, which is represented by the tuple (0,0).

In the 2-dimensional space, a point can be defined by a pair of values that correspond to the horizontal coordinate (x) and the vertical coordinate (y). The space can be divided into 4 zones (quadrants): Q1, Q2, Q3, Q4. Whose single point of union is the point (0, 0). 

If a point is in Q1 both its x coordinate and the y are positive. Below are two links for you to familiarize yourself with Cartesian coordinate system and Euclidean distance.

https://en.wikipedia.org/wiki/Cartesian_coordinate_system

https://en.wikipedia.org/wiki/Euclidean_distance


```python
# The list of tuples represents where you hit the target.
shots = [(4, 5), (-0, 2), (4, 7), (1, -3), (3, -2), (4, 5),
          (3, 2), (5, 7), (-5, 7), (2, 2), (-4, 5), (0, -2),
          (-4, 7), (-1, 3), (-3, 2), (-4, -5), (-3, 2),
          (5, 7), (5, 7), (2, 2), (9, 9), (-8, -9)]
```

#### 1. The ten-fold champion of Nod Archery Competition, Axel Cajselius is famous for being awesome and humble. But also for hitting an arrow with another arrow. Did you get one?

```python
myflag = True
for i in shots:
    if shots.count(i) > 1:
        print("YES I DID")
        myflag = False
        break
if myflag:
    print("I DID NOT")
```

#### 2. Calculate how many arrows have fallen in each quadrant.

```python
q1 = []
q2 = []
q3 = []
q4 = []
noq = []

for i in shots:
    if i[0] == 0 or i[1] == 0:
        noq.append(i)
    elif i[0] > 0 and i[1] > 0:
        q1.append(i)
    elif i[0] < 0 and i[1] > 0:
        q2.append(i)
    elif i[0] < 0 and i[1] < 0:
        q3.append(i)
    elif i[0] > 0 and i[1] < 0:
        q4.append(i)
    else:
        print("YOUR CODE IS BROKEN PLS FIX")
    
print(q1)
print(q2)
print(q3)
print(q4)
print(noq)
```

#### 3. Find the arrow that was closest to the center.

```python
best_shot = 0
for i in shots:
    if best_shot == 0 or abs(i[0]) + abs(i[1]) < abs(best_shot[0]) + abs(best_shot[1]):
        best_shot = i
print(best_shot)
```

#### 4. If the target has a radius higher than 9, you missed the target and must go get it in forest. How many arrows do you have to go pick up in the forest?

```python
woods_trip_count = 0

for i in shots:
    if abs(i[0]) + abs(i[1]) > 9:
        woods_trip_count += 1

print(woods_trip_count)
```
