```python

``` 
keyword to break out of the loop

````python
# Code here
from random import randint
random_number = randint(1,10)
guesses_left = 3

while guesses_left > 0:
    guess = int(input(f"Guesse an number from 1 to 10 you have {guesses_left} guessesleft!"))
    guesses_left -= 1
    if guess == random_number:
        print("YOU WIN! ")
        break
    if guesses_left == 0:
        print(f"YOU LOSE!!! CORRECT NUMBER WAS {random_number}!")


````

# Q12 - Help the snail calculate number of days for its escape!
A snail falls at the bottom of a 125 cm well. 

Each day the snail rises 30 cm. But at night, while sleeping, slides 20 cm because the walls are wet. 

How many days does it take to escape from the well?

````python
# Assign problem data to variables with representative names...
# ... such as: well height, daily advance, night retreat, accumulated distance
height = 135
day = 30
night = -20

# Assign 0 to the variable that represents the solution
progress = 0
days_passed = 0
# Write the code that solves the problem
while progress < height:
    progress += day
    if progress > height:
        print(f"ESCAPED IN {days_passed} DAYS!")
    else:
        days_passed += 1 
        progress += night




# Print the result with print('Days =', days)

````
