# Rock, Paper, Scissors

```python
# Import the choice function of the random module
# https://stackoverflow.com/questions/306400/how-to-randomly-select-an-item-from-a-list

import random
import math

# Assign to a list the 3 possible options: 'stone', 'paper' or 'scissors'.

options = ["rock", "paper", "scissors"]
best_of = 5
rounds_for_win = int(math.ceil((best_of / 2)))
isgameon = True
number_of_humans = 0
current_round = 0
player_1_score = 0
player_2_score = 0
# Define a function that randomly returns one of the 3 options.
# This will correspond to the play of the machine. Totally random.

def human_players():
    global isgameon
    user_input = input("HOW MANY HUMANS ARE PLAYING? 0 1 OR 2 ? (PRESS Q TO QUIT)")

    if user_input == "0" or user_input == "1" or user_input == "2":
        return user_input

    elif user_input.lower() == "q":      
        isgameon = False
        return -1

    else: human_players()


def get_choice_pc():
    return options[random.randint(0,2)]

# Define a function that asks your choice: 'stone', 'paper' or 'scissors'
# you should only allow one of the 3 options. This is defensive programming.
# If it is not stone, paper or scissors keep asking until it is.

def get_player_choice():
    global isgameon
    player_choice = input("Rock, paper, scissors, GO! \n What are you chosing? (Press H for help)")
    if player_choice.lower() == "rock" or player_choice == "paper" or player_choice == "scissors":
        return player_choice.lower()

    elif player_choice.lower() == "h":
        help_input = input("To play type rock paper or scissors. Type 'p' to continue the game play of any other key to quit")

        if help_input.lower() == "p":
            return get_player_choice()

        else:
            isgameon = False
            return "Quit"
    else:
        return get_player_choice()

# Define a function that resolves a combat.
# Returns 0 if there is a tie, 1 if the machine wins, 2 if the human player wins

def game_round(choice_p_1,choice_p_2):
    global player_1_score
    global player_2_score
    print(f"Player 1 chose {choice_p_1}, Player 2 chose {choice_p_2}")
    #TATKE IN ARGUMENTS AS (P1 CHOICE , P2 CHOICE)
    #RETURNS 0 FOR TIE 1 FOR P1 WIN 2 FOR P2 WIN
    if choice_p_1 == choice_p_2:
        print(f"You both chose {choice_p_1}")
        return 0

    elif choice_p_1 == "paper" and choice_p_2 == "rock":
        player_1_score += 1
        return 1

    elif choice_p_1 == "scissors" and choice_p_2 == "paper":
        player_1_score += 1
        return 1

    elif choice_p_1 == "rock" and choice_p_2 == "scissors":
        player_1_score += 1
        return 1

    else:
        player_2_score += 1
        return 2


    
# Define a function that shows the choice of each player and the state of the game
# This function should be used every time accumulated points are updated

def game_tally():
    global player_1_score
    global player_2_score
    global isgameon
    global rounds_for_win
    print("THE SCORE IS: \n Player1: " + player_1_score * "●" + (rounds_for_win - player_1_score) * "◯"  + "\n Player2: " + player_2_score * "●" + (rounds_for_win - player_2_score) * "◯" )

    if player_1_score == rounds_for_win:
        print("PLAYER 1 WON!!!")
        isgameon = False

    elif player_2_score == rounds_for_win :
        print("PLAYER 2 WON!!!")
        isgameon = False
    return

    
# Create two variables that accumulate the wins of each participant

def game (nr_players):
    print(f"GAME ON TILL THE BEST OF {best_of}")

    while isgameon:
        if nr_players == 0:
            game_round(get_choice_pc(),get_choice_pc())

        elif nr_players == 1:
            game_round(get_choice_pc(),get_player_choice())

        else:
            game_round(get_player_choice(),get_player_choice())
        game_tally()

    return
            

game(int(human_players()))


# Create a loop that iterates while no player reaches the minimum of wins
# necessary to win. Inside the loop solves the play of the
# machine and ask the player's. Compare them and update the value of the variables
# that accumulate the wins of each participant.


# Print by console the winner of the game based on who has more accumulated wins


    
```
