Tuples are similar to listst with the exception that tuples are *immutable*. Because of their similarity you can interact with tuple-objects in much the same way as you interact with list-objects.

# When to use Tuples

Why bother using tuples when they have fewer available methods? Frankly, tuples are not used nearly as often as lists in programming. But they come in handy when immutability is a priority. If you are passing around an object in your program  and need to make sure it does not get changed, then a tuple is your friend. It provides a convenient source of data integrity.

````python
done = ("Good", "Job")

print(done)
````
