# Q1

In the cell below, create a list named `random_list` with 50 unique random values. 

Requirements:

* Each value is an integer between 0 and 75.
* Each value in the list is unique.

*Tip:* Import the `random` library and use the `random.sample()` method to efficiently create your list with unique values ([more info here](https://docs.python.org/3/library/random.html#random.sample))

Print `random_list` to review its values

Print the length of `random_list`

```python
import random
random.sample(range(70),50)
random_list = random.sample(range(70),50)

print(random_list)
print(len(random_list))
print(len(set(random_list)))


```

## Q1.2 
Convert `random_list` to a set called `set1`. Print the length of the set to see if the length still is 50?

```python
# Solution here
set1 = set(random_list)
```

# Q2
Create an empty list named `random_list2` 

Your goal is to fill `random_list2` with 50 random numbers

Each value should be an integer falling between 0 and 75

Now we don't want only unique numbers so here we will **not** use the `sample()` method

*Tip:* Use a `for` loop, `append` and the `randint` function to solve this problem

```python
import random

random_list2 = []

while len(random_list2) < 50:
    random_list2.append(random.randint(1,75))

print(random_list2)

```

## Q2.1
Convert `random_list2` to a set called `set2`. Print the length of the set2. Is its length still 50?

```python
set2 = set(random_list2)
print(len(set2))
```

```python
# Explain why the length is different

# Answer: cause sets can only have unique values and duplicates dont get added to the set therefore set is smaller than the list
```
