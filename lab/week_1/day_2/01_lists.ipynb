# Q1 - Indian cuisine


Below you have a list of spices that are frequently used in the Indian cuisine.

```python
spice_lst = ["cardamom","fennel","garam masala","tumeric","cumin","coriander","mustard seeds","anise","cloves"]
```

## Q1.1
Black pepper is also a common spice in the Indian cuisine but it is missing in `spice_lst`. Add "black pepper" to `spice_lst`

```python
spice_lst.append("black pepper")
```

## Q1.2

You find it disturbing that the person who created this list did not list the spices in alphabetical order. Sort the spices alphabetically.

```python
spice_lst.sort()
print(spice_lst)
```

## Q1.3
Create a new list `c_lst` that only contains spices starting with the letter C. Create `c_lst` from `spice_lst`.

```python
c_list = []

for i in spice_lst:

    if i[0].lower() == "c":
        c_list.append(i)

print(c_list)
```

## Q1.4
Grab the `"mustard seeds"` and the `"tumeric"` from the `spice_lst`

```python
spice_lst.remove("mustard seeds")
spice_lst.remove("tumeric")
print(spice_lst)
```

## Q1.5
You want to spice thing up a little bit so you decide to add some fresh basil to the `spice_lst`.

```python
spice_lst.append("fresh basil")
```

## Q1.6
Damn, you just remembered that basil is not used in the Indian cuisine that much. Therefore, use a list method to get rid of the last item in `spice_lst`.

```python
spice_lst.pop()
```

# Q2 - Chocolate balls

You are making Chocolate balls, a Swedish delicacy. They are just like Meat balls except they are made out of chocolate instead of meat.

Your assignment is to take out the ingredients from your `pantry`. Your `pantry` is represented as a nested list (a list of lists). The 4 lists inside the big list represents the shelves in your pantry.

Grab the following ingredients from the `pantry` (using indexing) and add them to a new list called `choc_lst`:

    1.) Butter
    2.) Sugar
    3.) Vanilla Sugar
    4.) Cocoa
    5.) Oats
    6.) Coffee
    7.) Coconut flakes

```python
pantry = [["Pasta","Cocoa","Oats","Flour","Coconut Milk"],
          ["Rice", "Coffee", "Sambal Oelek", "Vanilla Sugar"],
          ["Coconut flakes", "Apple", "Mango", "Potato", "Butter"],
          ["Dates","Olives", "Sugar", "Canned Tomatoes"]]
```

```python
choc_lst = []
choc_lst.append(pantry[2][-1])
choc_lst.append(pantry[-1][2])
choc_lst.append(pantry[1][-1])
choc_lst.append(pantry[0][1])
choc_lst.append(pantry[0][2])
choc_lst.append(pantry[1][1])
choc_lst.append(pantry[2][0])
print(choc_lst)
```
