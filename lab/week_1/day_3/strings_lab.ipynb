# Q2
The list below is a recipe containing all ingredients to make the French dish ratatouille. Using this list you want to create a new list that you can call `grocery_list` 

The only ingredients you need to purchase are the ones starting with the letter `f`or the letter `y`.

Every item/ingredient in `grocery_list` should start with a capital letter

```python
ratatouille_recipe = ["Eggplants", "tomatoes", "yellow squashes", "zucchinis", "Olive oil", "onion", 
                "garlic", "red bell pepper","yellow bell pepper", "salt", "pepper", "crushed tomatoes",
               "fresh basil", "fresh parsley", "Fresh thyme"]
```

# Q5 - Blacklist!
In this assignment, your objective is to extract all the words from the poem except for the words that appear in the blacklist

But before we can match the content in our string-variable `poem` we need to do some cleaning!

1) remove all characters that are not in the alphabet\
2) lowercase all words

Now create a list storing all the words that are in poem except for the words in `blacklist`

Now remove any duplicate word from your list 

```python
blacklist = ['and', 'as', 'an', 'a', 'the', 'in', 'it']

poem = """No man is an island, entire of itself;
every man is a piece of the continent,
a part of the main.

If a clod be washed away by the sea,
Europe is the less,
as well as if a promontory were,
as well as if a manor of thy friend’s
or of thine own were.

Any man’s death diminishes me,
because I am involved in mankind;
and therefore never send to know
for whom the bell tolls;
it tolls for thee."""

```

```python
for i in poem:
    if not i.isalpha() and not i.isspace() and i != "\n" :
        poem = poem.replace(i,"")
poem = poem.lower()
    
poem_word_list = []

for i in poem.split():
    if i not in blacklist:
        poem_word_list.append(i)

print(poem_word_list)
print(len(poem_word_list))
```

```python
def no_reapeats(my_list):

    for i in my_list:


        if my_list.count(i) > 1:

            my_list.pop(my_list.index(i))
            no_reapeats(my_list)

    return my_list
        
poem_word_list = no_reapeats(poem_word_list)
print(poem_word_list)
```
