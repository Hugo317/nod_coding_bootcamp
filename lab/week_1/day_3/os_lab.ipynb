## 2nd Assignment: Understand recursion and directory traversal using os.

Task:
Write a Python function called `print_tree(path, depth=0)` that prints the structure of a directory and all its subdirectories in a tree-like format.

#### Requirements
The function should:

* Take a directory path as input

* Print all files and folders inside it

* Subdirectories should be printed recursively

* Each level deeper in the directory tree should be indented by two spaces

* Files and folders must be printed in the order returned by `os.listdir()`

* Assume the path exists and is a directory


<br>
<details>
<summary><b>Press for hint on Recursion</b> </summary>
Think of the function as a set of instructions for one single folder.

When the function finds another folder inside that folder, it needs to use the same instructions again — but for the new folder, and one level deeper.

That’s why the function must call itself inside the function, this what we mean by recursion.
Each self-call handles one deeper level of the directory tree, until there are no more folders to explore.

You are not looping “forever” — each call moves deeper into a smaller part of the problem, and eventually there are no subfolders left.
</details>


````python
path = "/Users/hugocarvalho/Documents/"


def computer_inspector(path):
    file_counter = 0
    dir_counter = 0
    for i in os.listdir(path):

        if os.path.isfile(os.path.join(path,i)):
            file_counter += 1 

        else:
            dir_counter += 1 
            sub_dirs, sub_files = computer_inspector(os.path.join(path,i))
            dir_counter += sub_dirs
            file_counter += sub_files

    return dir_counter, file_counter

    
nr_dirs, nr_files = computer_inspector(path)
print(f"Files: {nr_files}\nDirectories: {nr_dirs}")
````

### Q2. Find large files

Print the full path of all files that are larger than X MB. (X should be an int and stored as a variable).

````python
path = "/Users/hugocarvalho/Documents/"

too_big = 5

# path = "/Users/hugocarvalho/Downloads/"
for root, dirs, files in os.walk(path):
    for i in files:
        current_path = os.path.join(root,i)
        size_mb = os.path.getsize(current_path) / (1024 * 1024)
        if size_mb > too_big:
            print(root, dirs, i)

````
