```python
import pandas as pd
```

```python
df = pd.read_csv("titanic_imp.csv")
```

## 2. Handle inconsistet and/or irrelevant data

```python
df
```

```python
df["Fare"] = df["Fare"].str.replace("$","")
```

```python
df.Fare.head()
```

```python
df.describe(include="O")
```

```python
df.Survived.unique()
```

```python
#df.Survived.replace(to_replace=["yes","no"], value = ["1","0"],inplace=True)
df.replace({"Survived": {"yes": 1, "no": 0}}, inplace=True)
```

```python
df.Survived.unique()
```

## 3. Changing the data type with astype() 


```python
df.Survived = df.Survived.astype("int")
df.Fare = df.Fare.astype("float")
df.info()

```

```python
df.Age.value_counts(dropna=False)
```

```python
import numpy as np
```

```python
df.replace({"Age":{"Missing Data": np.nan}},inplace=True)
```

```python
df.info()
```

## Remove missing values with `dropna()`

```python
df.dropna()
```

```python
df.dropna(thresh=8)
```

```python
df.Age = df.Age.astype("float")
```

```python
df.Age = df.Age.fillna(df.Age.mean())
```

```python
df.Emb.mode()
```

```python
df.replace({"Emb":{np.nan: df.Emb.mode()[0]}})
```

```python
df.drop(columns="Deck",inplace=True)
```

```python
df.columns
```
