# comparison operators / evaluation methods

When filtering you can use any operation/expression/method that evaluates to True/False.

We will look at the following commonly used expressions/methods when filtering:


- `==`
- `<`
- `between()`
- `isin()`

```python
df.loc[df.Year < 2000]
```

```python
df.loc[df["City"].isin(["London","Year"])]
```

```python
df.loc[df["Year"].between(1992,2000)]
```

```python
#We must use parethesis whe using multiple conditions
df.loc[(df.Country == "SWE") & (df.Gender == "Women")]
```

```python
female = df.Gender == "Women"
```

```python
df.loc[swe & female]
```

```python
df["Country"].unique()
```

```python
df.loc[~(df["Gender"] == "Men") & df["Country"].isin(["SWE","NOR","DEN"])]
```
