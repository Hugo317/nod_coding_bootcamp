```python
# Lists in Python are versatile and can be used to store a collection of items.
# They are ordered, changeable (mutable), and allow duplicate elements.

# Use Case

# Creating a list of tasks
tasks = ["Read a book", "Write a report", "Attend a meeting"]

# Adding a new task
tasks.append("Prepare presentation")

# Removing a task
tasks.remove("Read a book")

# Accessing a task
print(tasks[1])  # Output: Write a report
```

```python
# Tuples are similar to lists but are immutable (cannot be changed after creation).
# They are useful for storing a collection of items that should not be modified.

# Use Case: Storing coordinates in a mapping application.

# Creating a tuple for coordinates (latitude, longitude)
location = (40.7128, -74.0060)

# Accessing elements
latitude, longitude = location
print(f"Latitude: {latitude}, Longitude: {longitude}")
```

```python
# Dictionaries store data in key-value pairs and are perfect for retrieving a value quickly by a known key.

# Use Case: Storing and retrieving user information in a web application.
# Creating a dictionary for user info
user = {"name": "Andrei", "age": 30, "email": "andrei@adventure.com"}

# Adding a new key-value pair
user["is_admin"] = True

# Accessing user's email
print(user["email"])
```

```python
# SETS
# a set is a collection data type that is unordered, mutable, and used to store unique elements.
# Sets are particularly useful for operations involving checking for membership,
# removing duplicates from a sequence, and performing mathematical operations like unions,
# intersections, and differences.


# Managing User Interests in a Social Media Application

# Creating sets for user interests
user1_interests = {"travel", "food", "photography", "music"}
user2_interests = {"books", "travel", "food", "yoga"}

# Adding a new interest
user1_interests.add("gardening")

# Removing an interest
user2_interests.discard("books")

# Finding common interests between two users
common_interests = user1_interests.intersection(user2_interests)
print(f"Common interests: {common_interests}")
# Output might be: Common interests: {'travel', 'food'}

# Checking if a user is interested in a specific topic
is_interested = "music" in user1_interests
print(f"Is user interested in music? {is_interested}")
# Output: Is user interested in music? True
```

### nested lists

A great feature of of Python data structures is that they support *nesting*. This means we can have data structures within data structures. For example: A list inside a list.

Let's see how this works!

```python
# Let's make three lists
lst_1=[1,2,3]
lst_2=[4,5,6]
lst_3=[7,8,9]

# Make a list of lists to form a matrix
matrix = [lst_1,lst_2,lst_3]
```

```python
#GRABBING NUMBER 8
matrix[2][1]
#GETTING NUMBER 4 AND 5 
matrix[1][:2]
#GETTING 7 AND 9 
matrix[2][::2]
```

```python
for i in matrix:
    for j in i :
        print(j**2)
```

```python
def min_max_avg(numbers):
    return min(numbers), max(numbers), sum(numbers)/len(numbers)
```
