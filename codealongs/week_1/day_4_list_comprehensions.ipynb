## LIST COMPREHENSIONS
List comprehensions is a way of combining storage and iteration streamlined logic!

```python
numbers = [1, 4, 9, 16, 25, 36, 49, 64, 81, 100]
```

```python
squared = []

for i in numbers:
    squared.append(i ** 2)

squared
```

```python
squared2 = [i ** 2 for i in numbers]
squared2
```

#NESTED LIST COMPREENHSIONS - TWO APROACHES


```python
matrix = [[1, 2, 3],
          [4, 5, 6],
          [7, 8, 9]]

squared_matrix =[[number ** 2 for number in row] for row in matrix]

squared_matrix


## BIG BRAIN SHIIIIIT YOOOOOOO
## STUDY THIS SNALLLAAAAA

# In this syntax, the outer loop iterates through each `row` in the matrix, and the inner loop 
# iterates through each `num` in the current `row`. This results in creating a new matrix where 
# each element is the square of the corresponding element in the original matrix.
```

```python
[num ** 2 for row in matrix for num in row] # with this aproach its a sigle list 
```

## List comp with long if statments

```python
applicants = [
    {'name': 'Alice', 'primary_coding_language': ["Python"], 'skills': ['Python', 'JavaScript', 'HTML', 'CSS']},
    {'name': 'Bob', 'primary_coding_language': ["Java"], 'skills': ['Java', 'Python', 'SQL']},
    {'name': 'Carol', 'primary_coding_language': ["Python"], 'skills': ['Python', 'Ruby', 'React']},
    {'name': 'David', 'primary_coding_language': ["Python"], 'skills': ['Python', 'Django', 'Angular']},
    {'name': 'Jean', 'primary_coding_language': ["Python"], 'skills': ['Python', 'JavaScript', 'HTML', 'CSS']},
    {'name': 'Claude', 'primary_coding_language': ["Python"], 'skills': ['Java', 'C++', 'SQL']},
    {'name': 'Finn', 'primary_coding_language': ["C#"], 'skills': ['C#', 'Ruby', 'React']},
    {'name': 'Achill', 'primary_coding_language': ["Python"], 'skills': ['Python', 'Django', 'Angular']}
]
```

```python
# we want to find a guy that has python as primary language and HTML and CSS skills 

[i["name"] for i in applicants if i["primary_coding_language"][0] \
 == "Python" and all(j in i["skills"] for j in ["HTML","CSS"])]

 
```

## Practical Uses for List Comprehensions

#### Reading Multiple Files
One use case where list comprehensions come in handy is when data is split across multiple files. For example, suppose we had a data directory that contained several CSV files (among other files), each with the same information (columns) for separate groups or divisions. We could use a list comprehension with an endswith('.csv') condition in it to get a list of just the CSV files in that directory. We could use another list comprehension to have Pandas read each of those files and then the pd.concat method to combine them all into a single data set that we can analyze as follows.

```python
import pandas as pd
import os
os.getcwd()
```

```python
path = os.path.join(os.getcwd(), "data")

path
```

```python
os.getcwd()
```

```python
os.listdir(path)
```

```python
[file for file in os.listdir(path) if ".xlsx" in file]
```

```python
## READING MULTIPLE FILES WITH LIST COMPRENSION
lst_files = [file for file in os.listdir(path) if file.startswith("boy") and file.endswith(".csv")]
```

```python
read_files = [pd.read_csv(os.path.join(path, file)) for file in lst_files]
read_files
```

```python
##DF IS CONVENTION FOR DATA FRAME
df = pd.concat(read_files, axis=0)
df
```

```python
df = pd.read_csv(os.path.join(path,"vehicles.csv"))
df
```
