## What Is a Jupyter Notebook?

A **Jupyter Notebook** is an interactive document where you can write and run Python code in small chunks, see the results immediately, and mix in notes, explanations, and visualizations. Instead of writing an entire program and running it all at once, you run pieces of code one at a time and watch what happens.

This makes notebooks especially friendly for beginners, data analysis, and experimentation.

## The Core Idea: Cells

A notebook is made up of **cells**. Each cell is a box you type into. There are two main types:

- **Code cells** – contain Python code you can run. The output (numbers, text, charts, errors) appears directly below the cell.
- **Markdown cells** – contain formatted text (headings, lists, links, notes). These are for explanations, not code.
You run a cell to execute it. When you run a code cell, Python processes whatever is inside and shows the result.

## How Code Runs: The Kernel

Behind every notebook is a **kernel** — the engine that actually runs your Python code and remembers everything. Two key things to understand:

1. **State is shared across cells.** If you create a variable in one cell, you can use it in another cell later. The kernel remembers it.
2. **Order matters.** Cells run in the order *you* run them, not necessarily top to bottom. Running cells out of order is the most common source of beginner confusion. When in doubt, restart the kernel and run everything from the top.
## The Two Modes

Notebooks have two modes, and knowing which one you're in is essential for keyboard shortcuts:

- **Edit mode** – you're typing inside a cell. The cell border is usually green or highlighted. Press `Enter` to enter edit mode.
- **Command mode** – you're acting on cells as a whole (adding, deleting, moving them). Press `Esc` to enter command mode.
Most keyboard shortcuts only work in **command mode**, so press `Esc` first if a shortcut isn't working.

## A Typical Workflow

1. Type some code into a code cell.
2. Run it and check the output.
3. Add a new cell below to continue building on what you have.
4. Use markdown cells to explain your thinking as you go.
5. If things get messy, restart the kernel and run all cells from the top to confirm everything works in order.

## Quick Tips for Beginners

- If something behaves strangely, **restart the kernel and run all cells** to get a clean slate.
- Keep cells **small and focused** — one idea per cell makes debugging easier.
- Run cells **top to bottom** when you're done experimenting, to make sure your notebook works in order.
- Press `Esc` if a keyboard shortcut doesn't seem to work — you're probably in edit mode.

## Pattern to solve many problems in Python

When writing code, we often find ourselves working with large amounts of data in lists, dictionaries, or other structures. We want to perform some operation on each item in that data, so we can use a **for loop** to iterate over the data and apply our logic to each item. This pattern of "looping over data and applying logic" is fundamental in programming and can be used to solve a wide variety of problems.

But most of the time, we don't want to just print the results; we want to **store** the results in a new list or dictionary. This is where the pattern of "creating an empty list/dictionary, looping over data, applying logic, and appending results" comes in. This pattern allows us to build up new data structures based on our logic applied to existing data.

```python
# In this example, we have a list of ages and we want to save the ones who are allowed to buy wine at Systembolaget.

ages = [25, 42, 17, 30, 55, 12, 40, 22, 18, 60, 45, 33, 27, 19, 50, 38, 29, 24, 31, 41, 26, 33,
        27, 28, 34, 36, 21, 44, 39, 32, 37, 23, 26, 30, 29, 40, 31, 27, 35, 28, 33, 24]
```

## `While Loops`


The `while` statement in Python is one of most general ways to perform iteration. A <code>while</code> statement will repeatedly execute a single statement or group of statements as long as the condition is true. The reason it is called a 'loop' is because the code statements are looped through over and over again until the condition is no longer met.

The general format of a while loop is:

    while test:
        code statements
    else:
        final code statements

Let's look at a few simple <code>while</code> loops in action.

```python
x = 0
while x < 10:
    print(f"'x' is currently: {x}")
    print("\t'x' is still less than 10, adding 1 to 'x'")
    x += 1
```

## `break` & `continue`

We can use <code>break</code>, <code>continue</code> statements in our loops to add additional functionality for various cases. The three statements are defined by:

    break: Breaks out of the current closest enclosing loop.
    continue: Goes to the top of the closest enclosing loop.
    pass: Does nothing at all.


Thinking about <code>break</code> and <code>continue</code> statements, the general format of the <code>while</code> loop looks like this:

    while test:
        code statement
        if test:
            break
        if test:
            continue
    else:

<code>break</code> and <code>continue</code> statements can appear anywhere inside the loop's body, but we will usually put them further nested in conjunction with an <code>if</code> statement to perform an action based on some condition.

<img src="https://i.stack.imgur.com/Btr3L.png" width="400" height="500">

```python
x = 0
while x < 10:
    if x == 5:
        print("'x' is equal to", x)
        break
    print(f"'x' is currently: {x}")
    print("\t'x' is still less than 10, adding 1 to 'x'")
    x += 1
```

```python
x = 0
while x < 10:
    if x == 5:
        print("'x' is equal to", x)
        print("continuing")
        continue
    print(f"'x' is currently: {x}")
    print("\t'x' is still less than 10, adding 1 to 'x'")
    x += 1
```
