# Introduction to User Defined Functions
In our journey through Python programming, we've primarily written code in a procedural manner. We've used functions and methods from Python's standard library or other libraries like Numpy or Pandas. This approach has served us well, but to advance from being a beginner to an intermediate Python programmer, it's essential to grasp the fundamentals of functional programming and learn how to craft your own functions.

##  Understanding Procedural Programming
Procedural programming involves writing code line by line and executing it in the order it's written. It's often the starting point for many aspiring programmers because it follows a straightforward sequence of actions to solve a problem: do this, then do that, and so on.

## Embracing Functional Programming
More experienced programmers tend to approach problem-solving differently. They adopt a functional programming approach, where they create functions that consist of multiple lines of code to accomplish specific tasks. These functions often return results that can be utilized by other functions.

Think of a function as a tool – something that aids you in accomplishing a task. Functional programming empowers you to craft your own tools. Some of these tools will be tailored to the immediate tasks at hand, while others will be more versatile and reusable across various projects. Picture a chef who can craft their perfect set of knives or a carpenter who can create their ideal set of saws, perfectly suited for their specific needs. That's the power that functional programming offers to programmers. Over time, you'll build a personal library of functions that enhance your efficiency and effectiveness in your work.

```python
def categorize_expenses(expense_dict):
    rent, gorceries, utilities, entertainment = expense_dict["rent"], expense_dict["groceries"], expense_dict["utilities"], expense_dict["entertainment"]

    total_expenses = calculate_monthly_expenses(rent,groceries,utilities,entertainment)

    essential = (rent , groceries , utilities)
    non_essential = (entertainment)

    return total_expenses, essential, non_essential

expenses = {
    "rent": 12000,
    "groceries": 4000,
    "utilities": 2000,
    "entertainment": 1000
}

total,essential,non_essential = categorize_expenses(expenses)

print(f"total: {total}  essential: {essential},  non-essential: {non_essential}")
```

# Namespaces are honking good idea - let's do more of those! 


The Local, Global, Enclosing, and Built-in (LEGB) rule is used to determine the scope and namespace of a variable in Python. Let's create three functions to illustrate these concepts:

The quote "Namespaces are one honking good idea - and why we should do more of those" is from the Zen of Python, which is a collection of guiding aphorisms for writing computer programs in the Python language. This particular aphorism emphasizes the importance of using namespaces and encapsulation to manage and organize code effectively.

Here's a brief explanation of why namespaces are considered a good idea:

Organization: Namespaces allow you to organize your code into logical units, preventing naming conflicts. They act like containers for variables, functions, and classes, providing a way to group related code together.

Modularity: Namespaces promote modularity, which is a key software engineering principle. By dividing your code into namespaces, you can build complex systems by combining smaller, well-encapsulated components. This enhances code readability and maintainability.

Avoiding Conflicts: Namespaces prevent naming conflicts by ensuring that two entities with the same name in different namespaces don't interfere with each other. This is crucial when working on large projects with multiple contributors.

Clarity: Namespaces make your code more readable and self-explanatory. When you see a variable or function prefixed with a namespace, it's clear where it comes from and how it relates to the rest of the code.

Overall, the quote encourages developers to embrace the concept of namespaces and apply it more extensively in their code to improve organization, modularity, and code clarity. It's a reminder of good software engineering practices and the benefits of structured, well-organized code.

```python
x = "I AM GLOBAL"

def local_scope():
    x = "I am a local variable"
    return

def enclosing_scope():
    x = "I am an enclosing var"

    def inner_function():
        x = "Local rules"
        return x
    return inner_function(), x 

def global_scope():
    return x

print(enclosing_scope())
print(global_scope())
```
