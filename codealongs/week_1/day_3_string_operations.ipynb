```python
# List of strings to demonstrate string methods
strings_to_check = [
    "Hello123",
    "12345",
    "abcDEF",
    "   ",
    "PythonIsFun",
    "is_this_fun",
    "First We Take Manhattan",
    "CAPSLOCK",
    "CAPS LOCK"
]


# Loop over the list of strings
for string in strings_to_check:
    print(f"String: '{string}'")
    if string.isalnum():
        print("\tisalnum(): The string consists of alphanumeric characters.")
    if string.isalpha():
        print("\tisalpha(): The string consists of alphabetic characters.")
    if string.isdigit():
        print("\tisdigit(): The string consists of numeric characters.")
    if string.islower():
        print("\tislower(): The string is in lowercase.")
    if string.isupper():
        print("\tisupper(): The string is in uppercase.")
    if string.istitle():
        print("\tistitle(): The string is titlecased.")
    if string.isspace():
        print("\tisspace(): The string consists of whitespace characters.")

    print()
    print()
```
