 ## `map` function 

```python
# standard function that accepts at least two arguments, a function and an iterable / sequence

def farhenheit(celsisus):
    return (9/5)* celsisus + 32

temps = [0, 22.5, 40, 100]

f_temps = list(map(farhenheit,temps))

f_temps
```

```python
list(map(lambda x: (9/5) * x +32, temps))
```

```python
a = [1,2,3,4]
b = [5,6,7,8]
c = [9,10,11,12]

list(map(lambda x,y,z: x+y+z, a,b,c))
```

`filter`

```python
# Also a standard function in python

# There is a lambda for each value in the iterable.

# Returns filter object which can be converted into other iterables

# The object contains only the values that return true to the lambda
```

```python
users = [
    {"username": "the_realdonaltrump", "tweets": ["We are going to build a wall"]},
    {"username": "hillary_clint0n", "tweets": ["Our assessment is that the Egyptian government is stable and is looking for ways to respond to the legitimate needs and interests of the Egyptian people."]},
    {"username": "doofus", "tweets": []},
    {"username": "splooden", "tweets": []},
    {"username": "port_side", "tweets": ["I love sailing"]},
    {"username": "cadmium_boi", "tweets": []},
    {"username": "flicker", "tweets": ["Neymar is so OP, duuude"]},
    {"username": "carl_bildt", "tweets": ["Pyramidalt klumpigt av regeringen att inte tvinga allt bära ansiktsmask"]}
]
```

```python
inactive_uresrs = list(filter(lambda u: len(u["tweets"]) == 0, users))

inactive_uresrs
```

```python
inactive_uresrs = list(filter(lambda u: not u["tweets"], users))
inactive_uresrs
```

```python
#combining map and filter!

#mapping only users
list(map(lambda user: user["username"], users))
```

`reduce`

```python
# nota standard function anymore, you need to import it 

from functools import reduce
```

```python
lst = [1,2,3,4,5]
prod = reduce((lambda x,y: x*y), lst)

prod
```

```python
lst = [47, 11, 42,13]

#ALWAYS HAVE TO USE 2 ELEMENTS TO OPERATE ON EACHOTHER
reduce(lambda x,y: x+y, lst)
```

## `zip`

```python
# Make an iterator that aggregates elements from each of the iterables.

# Returns an iterator of tuples, where the i-th tuple contains the i-th element from each of
# the argument sequences or iterables.

# The iterator stops when the shortest input iterable is exhausted.
nums = [1,2,3,4,5,6,7]
words = ["one", "two", "three", "four","five"]

list(zip(nums,words))
#WILL NOT COPY EXTRA INDEXES WILL NOT GIVE AN ERROR 
```

```python
## zipping more than 2 lists

haha = ["AHAH","LOL","funny","mdr","lmao"]
list(zip(nums,words,haha))
```

```python
stock_prices = [("STAB",234),("SEB",97),("KINV",163),("ASSA",247)]

list (zip(*stock_prices))
```

```python
ticker, price = list (zip(*stock_prices))
print(ticker,price)
```

```python
# test score from two smesters, the final grade is equal to the score of the best test result

score1 = [17,14,16]
score2 = [12,18,15]
students = ["Knatte","Fnatte","Tjatte"]

[max(pair) for pair in zip(score1, score2)]

{student: max(test1,test2)for student, test1,test2 in zip(students, score1, score2)}
```

## `enumerate`

```python
# very usefull function when we want to keep track of the index 
lst = ["Sweden","Norway","Denmark","Finland","Iceland","Faroe Islands"]

for index, country in enumerate(lst,start=1):
    print(index, country)
```
