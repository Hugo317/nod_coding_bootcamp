```python
def display_board(board):
    clear_output()

    print("   |   |")
    print(" " + board[1] + " | " + board[2] + " | " + board[3]  )
    print("   |   |")
    print("------------")
    print("   |   |")
    print(" " + board[4] + " | " + board[5] + " | " + board[6]  )
    print("   |   |")
    print("------------")
    print("   |   |")
    print(" " + board[7] + " | " + board[8] + " | " + board[9]  )
    print("   |   |")
test_board = ["#","X","O","X","X","O","X"," ","0","X"]
display_board(test_board)
```

```python
def player_input():
    """ Function will ask players if they want X or O in the start"""
    marker = ""
    while not (marker == "X" or marker == "O"):
        marker = input("Player 1: Do you want fo be X or O?").upper()

    if marker == "X":
        return ("X","O")
    else:
        return ("O","X")
    
```

```python
def win_check(board, mark):
    return( (board[1] == mark and board[2] == mark and board[3] == mark) or 
            (board[4] == mark and board[5] == mark and board[6] == mark) or
            (board[7] == mark and board[8] == mark and board[9] == mark) or
            (board[1] == mark and board[4] == mark and board[7] == mark) or
            (board[2] == mark and board[5] == mark and board[8] == mark) or
            (board[3] == mark and board[6] == mark and board[9] == mark) or
            (board[1] == mark and board[5] == mark and board[9] == mark) or
            (board[3] == mark and board[5] == mark and board[7] == mark)
            )
```

```python
test2_board = ["#","O","O","O","X","O","X"," ","X","X"]
def my_win_check(board):
    win_con_list = [[1,2,3],[4,5,6],[7,8,9],[1,4,7],[2,5,8],[3,6,9],[1,5,9],[3,5,7]]

    #iterates over in con list
    for i in win_con_list:

        #for each list use the values of indexes 0, 1, 2 to acess board on a winning line and turn into a set
        check = set([board[i[0]], board[i[1]], board[i[2]]])

        #if the set's length is 1 and its not a space we have a winner
        if len(check) == 1 and " " not in check:
            return True

    #if the loop does not find a winning line return false
    return False

print(my_win_check(test2_board))

#print(display_board(test2_board))

```

```python
import random
def choose_first():
    if random.randint(0,1) == 0:
        return "Player 1"
    else:
        return "Player 2"
```

```python
def space_check(board, position):
    return board[position] == " "
```

```python
def full_boar_check(board):
    for i in range(1,10):
        space_check(board,i)
        return False   
    return True
```

```python
def player_choice(board):
    position = 0

    while position not in range(1,10) or not space_check(board,position):
        position = int(input("choose your next position (1-9) "))

    return position
```

```python
def replay():
    return input("do you want to play again?").lower().startswith("y")
```

```python
def change_marker(marker):
    if marker == "X":
        return "O"
    else:
        return "X"
```

```python
def setup_game():
    board = [" "] * 10
    player1_marker, player2_marker = player_input()
    turn = choose_first()

    if turn == "Player 1":
        current_marker = player1_marker
    else:
        current_marker = player2_marker

    print(f"{turn} goes first!")

    return board, current_marker
```

```python
print("Welcome to Tic Tac Toe")

def play_tic_tac_toe():

    the_board, current_marker = setup_game()

    play_game = input("Are you ready to play? Y/N")

    if play_game.lower().startswith("y"):
        game_on = True

    else:
        game_on = False

    while game_on:
            display_board(the_board)
            place_marker(the_board, current_marker, player_choice(the_board))

            if my_win_check(the_board):
                game_on = False
                display_board(the_board)
                print("YOU WIN FREAK")
                break

            elif full_boar_check(the_board):
                display_board(the_board)
                print("TIC TAC TIE")
                game_on = False
                break
            else:
                current_marker = change_marker(current_marker)
    if replay():
        return play_tic_tac_toe()
```
